```python

# 📦 Phase 2: Simulated Kubernetes Agent (Google Colab)
import pandas as pd

# Load saved predictions from Phase 1
predictions_df = pd.read_csv("predictions.csv")
predictions_df.head()
```

```python

# ⚙️ Simulated Kubernetes Agent Logic

# For visualization: assign dummy pod names
predictions_df["Pod_Name"] = [f"pod-{i}" for i in range(len(predictions_df))]

# Detect issues
issue_rows = predictions_df[predictions_df["Predicted"] == 1]

if not issue_rows.empty:
    print("⚠️ Issues detected by the model:")
    for _, row in issue_rows.iterrows():
        pod_name = row["Pod_Name"]
        print(f"🔁 Simulated: Restarting pod '{pod_name}'")
else:
    print("✅ No issues detected by the agent.")
```

```python
import matplotlib.pyplot as plt

issue_count = predictions_df["Predicted"].value_counts().sort_index()
labels = ["No Issue", "Issue"]

plt.figure(figsize=(6, 4))
plt.bar(labels, issue_count)
plt.title("Prediction Summary by Agent")
plt.ylabel("Number of Pods")
plt.xlabel("Prediction Outcome")
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()

```

```python
# 📝 Save issue list locally (no download prompt)
if not issue_rows.empty:
    issue_rows.to_csv("agent_issue_log.csv", index=False)
    print("✅ Issue log saved as 'agent_issue_log.csv'")

```
